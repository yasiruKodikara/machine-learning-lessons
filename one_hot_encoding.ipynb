```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("/content/data_for_one_hot_enc.csv")
```

```python
data["result"]
```

```python
result_category = data["result"]
```

# Method_1


```python
from sklearn.preprocessing import LabelEncoder
obj = LabelEncoder()
result = obj.fit_transform(result_category)
```

```python
result
```

# Method_2 (One Hot Encoding)

```python
from sklearn.preprocessing import LabelBinarizer
obj_2 = LabelBinarizer()
result_2 = obj_2.fit_transform(result_category)
```
